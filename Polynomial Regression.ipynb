```python
# 5. Adjusted R²

r_squred = 0.94
features = [1,2,3]
n = 10
p = len(features)


Adjusted_r_squred = round(1-(((1- r_squred)*(n-1))/(n-p-1)),3)
print(f'Adjusted R² = {Adjusted_r_squred}')
```

# 8. What is Adjusted R²?
## question:
#### What is Adjusted R², and why do we need it for a model like this one, instead of just using R²?

## answer:
#### Adjusted $R^2$ penalizes models for adding unnecessary features, unlike standard $R^2$ which always increases. It only increases if a new feature meaningfully improves predictive accuracy beyond random chance. This mechanism prevents overfitting and ensures that complex models rely only on genuinely useful predictors.
