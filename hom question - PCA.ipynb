```python
# 16. Variance → projection → explained variance
def col_Variance(col_val):
    mean = sum(col_val) / len(col_val)
    variance = []
    for x in col_val:
        X_mean = (x-mean)**2
        variance.append(X_mean)
    return sum(variance)/len(col_val)


# (a) Compute the variance of each of the three columns

    ## col 1 average speed deviation (km/h):
speed_dev = [5,15,15,25]
speed_dev_variance = col_Variance(speed_dev)

    ## col 2 number of stops
numb_stops = [10,30,50,30]
numb_stops_variance = col_Variance(numb_stops)

    ## col 3  idle time (minutes):
idle_T = [5,15,15,5]
idle_T_variance = col_Variance(idle_T)

print('(a) Compute the variance of each of the three columns:\n')
print(f'    col 1 "average speed deviation (km/h)", variance: |  {speed_dev_variance}')
print(f'    col 2 "number of stops", variance:                |  {numb_stops_variance}')
print(f'    col 3 "idle time (minutes)", variance:            |  {idle_T_variance}')

```

```python
# (b) Add them up — that is the total variance of the dataset

tot_variance = numb_stops_variance + speed_dev_variance + idle_T_variance

print('(b) Add them up — that is the total variance of the dataset:')
print(f'    the total variance for the dataset is: {tot_variance}')
```
