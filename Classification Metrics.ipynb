```python
import pandas as pd

tot_patients = 250
tp = 45
fp = 15
fn = 25
tn = 165

Accuracy = (tp+tn)/tot_patients
print(F'Accuracy: {Accuracy}')

Precision = tp/(tp+fp)
print(F'Precision: {Precision}')

Recall  = tp/(tp + fn)
print(F'Recall: {round(Recall,3)}')

F1  = 2*((Precision * Recall)/(Precision + Recall))
print(F'F1: {round(F1,3)}')
```
