```python
x_vals = [1, 3, 5, 7]
y_vals = [19, 25, 41, 51]
n = len(x_vals)

# (a) pred for 9 H of learming
x_a = 9
y_pred_a = 6 * x_a + 10
print(f"(a) Predicted score (x={x_a}): {y_pred_a}")

# (b) residual for a student with x = 5
x = 5
y = 41
y_pred = 6 * x + 10
e = y - y_pred
print(f"(b) Residual (x={x}): {e}")

# (c) Completed residual, compute table MSE and RMSE
print("\n(c) Residuals Table:")
sse = 0
for i in range(n):
    y_pred_i = 6 * x_vals[i] + 10
    e_i = y_vals[i] - y_pred_i
    e_sq = e_i ** 2
    sse += e_sq
    print(f"x={x_vals[i]}, y={y_vals[i]} ---> y_pred={y_pred_i}, e={e_i}, e^2={e_sq}")

mse = sse / n
rmse = mse ** 0.5
print(f"\nSSE: {sse}")
print(f"MSE: {mse}")
print(f"RMSE: {round(rmse, 3)}")

# (d) Calc SST, R^2 + writen Explenation
y_mean = sum(y_vals) / n
sst = sum((y_i - y_mean) ** 2 for y_i in y_vals)
r_squared = 1 - (sse / sst)

print(f"\n(d) SST: {sst}")
print(f"R^2: {round(r_squared, 3)}")
print("Meaning: The model explains around 96.9% of the variance in the quiz scores.")
```
