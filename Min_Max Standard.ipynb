```python
#A. Min-max normalisation

numbers = [18,42,66,90]

max_num = max(numbers)
min_num = min(numbers)

for numb in numbers:
    x_norm = (numb-min_num)/(max_num-min_num)
    print(f'maped ---> to {numb} = {round(x_norm,2)}')
```

```python
# B. Standard deviation
# del std
number_2 = [8, 10, 10, 14, 14, 16]
len_numbs = len(number_2)
mean_num = sum(number_2) / len_numbs

print(f"Mean: {mean_num}") # הדפסת הממוצע

sum_squared_diffs = 0
for numb in number_2:
    x = numb - mean_num
    print(f"Deviation for {numb}: {x}\n----------------------") # הדפסת כל סטייה
    sum_squared_diffs += (x ** 2)

variance = sum_squared_diffs / len_numbs
std = variance ** 0.5
print(f"\nFinal Standard Deviation: {round(std,2)}")

    
```

## C. Comparing the two scalers:

### Min-Max Normalisation-
#### Advantage: Preserves the original distribution shape and scales all data into a fixed, predefined boundary, typically between 0 and 1
#### Disadvantage: Highly sensitive to outliers, as a single extreme value will compress the rest of the data into a very narrow range

### Standardisation
#### Advantage: Handles outliers much better than Min-Max scaling and is highly effective for algorithms that assume a normal distribution centered around zero, such as Support Vector Machines (SVMs) or logistic regression.
#### Disadvantage: Does not restrict the data to an absolute, bounded range, making it unsuitable for models that require strictly bound inputs.
