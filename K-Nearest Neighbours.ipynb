```python
# Question 11: Scale first, then measure the distance.
    
    # (a) Min-max normalise the tickets and usage

tikets = [2,9,6]
max_tickets = 10
min_tickets = 0

usage = [80,15,40]
max_usage = 100
min_usage = 0

tikets_normelized = []
usage_normelized = []

for numb in tikets:
    tikets_norm = (numb-min_tickets)/(max_tickets-min_tickets)
    tikets_normelized.append(round(tikets_norm,3))

for numb in usage:
    usage_norm = (numb-min_usage)/(max_usage-min_usage)
    usage_normelized.append(round(usage_norm,2))

print(f'tikets normalized = {tikets_normelized}')
print(f'usage normalized = {usage_normelized}')
```

### Handling Ties in KNN (e.g., k=4 splitting 2-2)

When a tie occurs in the K-Nearest Neighbors algorithm, there are several common strategies to break it:

* **Reduce k:** Ignore the farthest neighbor and re-evaluate the prediction using $k=3$. An odd value for $k$ naturally prevents a two-way tie.
* **Distance Weighting:** Assign higher weights to the votes of the neighbors that are physically closer to the new data point (typically weighting by $1/d$), rather than treating all neighbors equally.
* **Prior Probability (Class Distribution):** Break the tie by defaulting to the class that is more prevalent in the overall training dataset.
* **Random Selection:** Randomly choose between the tied classes (i.e., flip a coin).
